## Time complexity

Computer programs take time to run. How much time depends on the size of the task and on programming decisions. 

### $O(N)$

First consider a array of size $N$. We want to perform one operation on each element. An example of this is the addition of 2 vectors.


```python
import numpy as np
import time

# Add two vectors
def vector_add(A, B):
    # Only works if A and B are the same size
    result = np.zeros(A.size)
    for i in range(A.size):
        result[i] = A[i] + B[i]
    return result

# Measure time to add two vectors
def vec_add_time(A, B, N):
    start = time.time()
    C = vector_add(A, B)
    end = time.time()
    print(f"For two arrays of length {N}, the addition took {end-start} seconds")
    
# Test 10000 array
N = 10000
A = np.random.rand(N)
B = np.random.rand(N)
vec_add_time(A, B, N)

# Test 100000 array
N = 100000
A = np.random.rand(N)
B = np.random.rand(N)
vec_add_time(A, B, N)

# Test 1000000 array
N = 1000000
A = np.random.rand(N)
B = np.random.rand(N)
vec_add_time(A, B, N)
```

We can see that the time for each code scales somewhat linearly.

### $O(N^2)$

If instead, we look at adding two square *matrices* instead of vectors. We get:

```python
# Add two vectors
def matrix_add(A, B):
    # Only works if A and B are the same size
    result = np.zeros(A.shape)
    for i in range(A.shape[0]):
        for j in range(A.shape[1]):
            result[i,j] = A[i,j] + B[i,j]
    return result

# Measure time to add two vectors
def mat_add_time(A, B, N):
    start = time.time()
    C = matrix_add(A, B)
    end = time.time()
    print(f"For two {N}x{N} matrices, the addition took {end-start} seconds")
    
# Test 100x100 matrix
N = 100
A = np.random.rand(N, N)
B = np.random.rand(N, N)
mat_add_time(A, B, N)

# Test 1000x1000 matrix
N = 1000
A = np.random.rand(N, N)
B = np.random.rand(N, N)
mat_add_time(A, B, N)

# Test 10000x10000 matrix
N = 10000
A = np.random.rand(N, N)
B = np.random.rand(N, N)
mat_add_time(A, B, N)
```

Imagine a algorithm that requires you to multiply two square matrices of $N\times N$. 

What happens if we perform the multiplication "by hand"?

```python
# Function that multiplies 2 matrices
def matrix_mult(A, B):
    # Only works if A and B are of the correct size
    result = np.zeros((A.shape[0], B.shape[1]))
    for i in range(A.shape[0]):
        for j in range(A.shape[1]):
            for k in range(B.shape[1]):
                result[i,k] = result[i,k] + A[i,j]*B[j,k]
    return result

# Time the multiplication
def mult_time(A, B, N):
    start = time.time()
    C = matrix_mult(A, B)
    end = time.time()
    print(f"For {N}x{N} matrices, the multiplication took {end-start} seconds")

# Test 10x10 matrices
N = 10
A = np.random.rand(N,N)
B = np.random.rand(N,N)
mult_time(A, B, N)

# Test 100x100 matrices
N = 100
A = np.random.rand(N,N)
B = np.random.rand(N,N)
mult_time(A, B, N)

# Test 300x300 matrices
N = 300
A = np.random.rand(N,N)
B = np.random.rand(N,N)
mult_time(A, B, N)
```

This operation scales as $N^3$ because of the nested loop. We say that its time complexity is $O(N^3)$. When $N$ increases by a factor of $30$ ($10 \rightarrow 300$), $N^3$ increases by a factor of 27000 as can be seen by $0.0003\times 27000 \approx 8.1$, close to $7.25$ seconds. The math is probably wrong because of overhead that has more of an impact in the 10x10 example.
