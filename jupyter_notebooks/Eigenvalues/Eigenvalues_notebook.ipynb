# Eigenvalues notebook

Consider the real symmetric matrix:

$$
A = 
\begin{pmatrix}
5 & 1 & 7 & 3 & 0 \\
1 & 3 & 2 & 7 & 1 \\
7 & 2 & 1 & 8 & 2 \\
3 & 7 & 8 & 2 & 3 \\
0 & 1 & 2 & 3 & 4 
\end{pmatrix}
$$

There are a range of functions in the numpy and scipy libraries that can be used to find the eigenvalues and eigenvectors.

## numpy

```python
import numpy as np 

A = np.array([[5, 1, 7, 3, 0], [1, 3, 2, 7, 1], [7, 2, 1, 8, 2], [3, 7, 8, 2, 3], [0, 1, 2, 3, 4]], dtype='float')

# Find eigenvalues only
print(f"TEST 1:")
x1 = np.linalg.eigvalsh(A)
print(f"The eigenvalues of A are {x1}.")

# Find the eigenvalues and eigenvector
print(f"{'-':-<80}")
print(f"TEST 2:")
x2, v2 = np.linalg.eigh(A)
print(f"{"Eigenvalue":<20} Eigenvector ")
for i in range(5):
    print(f"{x2[i]:<20.8f} {v2[:,i]}")
```

## scipy

```python
import scipy as sp 

# Find eigenvalues only
print(f"TEST 1:")
x1 = sp.linalg.eigvalsh(A)
print(f"The eigenvalues of A are {x1}.")

# Find the eigenvalues and eigenvector
print(f"{'-':-<80}")
print(f"TEST 2:")
x2, v2 = sp.linalg.eigh(A)
print(f"{"Eigenvalue":<20} Eigenvector ")
for i in range(5):
    print(f"{x2[i]:<20.8f} {v2[:,i]}")
```

## Different matrices

The functions shown above can also find eigenvalues and eigenvectors for *Hermitian* matrices that are very common in physics. 

For more general matrices, use `eigval` and `eig` though the values will often be complex. 

## Comparison between numpy and scipy

For $A\vec{v} = \lambda \vec{v}$, both library work well. But scipy has more capabilities for more complicated problems.
